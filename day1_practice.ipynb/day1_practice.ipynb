```python
# Day 1: Latihan Logika Python Dasar - Simulasi Penjualan
diskon_ramadan = 0.50
diskon_lebaran = 0.85

harga_asli = 100000

# Hitung harga setelah diskon
harga_ramadan = harga_asli * (1 - diskon_ramadan)
harga_lebaran = harga_asli * (1 - diskon_lebaran)

print(f"Harga normal: Rp{harga_asli:,}")
print(f"Harga pas Ramadan (Diskon 50%): Rp{int(harga_ramadan):,}")
print(f"Harga pas Lebaran (Diskon 85%): Rp{int(harga_lebaran):,}")
```
